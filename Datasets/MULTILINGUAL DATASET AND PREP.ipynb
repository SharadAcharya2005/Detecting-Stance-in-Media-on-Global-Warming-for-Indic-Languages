```python
import pandas as pd

english_df = pd.read_csv("GWSD (1).tsv", sep="\t")

hindi_df = pd.read_csv("GWSD_Hindi.tsv", sep="\t")

bengali_df = pd.read_csv("GWSD_Bengali.tsv", sep="\t")

print("English:", english_df.shape)
print("Hindi:", hindi_df.shape)
print("Bengali:", bengali_df.shape)
```

```python
print("ENGLISH COLUMNS:")
print(english_df.columns.tolist())

print("\nHINDI COLUMNS:")
print(hindi_df.columns.tolist())

print("\nBENGALI COLUMNS:")
print(bengali_df.columns.tolist())
```

```python
print("English has guid:", "guid" in english_df.columns)
print("Hindi has guid:", "guid" in hindi_df.columns)
print("Bengali has guid:", "guid" in bengali_df.columns)
```

Check the translated column names

```python
print("Hindi columns:")
print(hindi_df.columns.tolist())

print("\nBengali columns:")
print(bengali_df.columns.tolist())
```

Check GUID matching

```python
print("English unique GUIDs:", english_df["guid"].nunique())
print("Hindi unique GUIDs:", hindi_df["guid"].nunique())
print("Bengali unique GUIDs:", bengali_df["guid"].nunique())

print("English-Hindi match:",
      set(english_df["guid"]) == set(hindi_df["guid"]))

print("English-Bengali match:",
      set(english_df["guid"]) == set(bengali_df["guid"]))
```

Merge all three datasets

```python
# Select only translated columns
hindi_translation = hindi_df[["guid", "sentence_hi"]]

bengali_translation = bengali_df[["guid", "sentence_bn"]]

# Start with original English dataset
multilingual_df = english_df.copy()

# Merge Hindi
multilingual_df = multilingual_df.merge(
    hindi_translation,
    on="guid",
    how="inner",
    validate="one_to_one"
)

# Merge Bengali
multilingual_df = multilingual_df.merge(
    bengali_translation,
    on="guid",
    how="inner",
    validate="one_to_one"
)

print("Final dataset shape:", multilingual_df.shape)
print(multilingual_df.head())
```

Verify the translations

```python
print("Missing English:",
      multilingual_df["sentence"].isna().sum())

print("Missing Hindi:",
      multilingual_df["sentence_hi"].isna().sum())

print("Missing Bengali:",
      multilingual_df["sentence_bn"].isna().sum())
```

Save our multilingual dataset

```python
multilingual_df.to_csv(
    "GWSD_Multilingual.tsv",
    sep="\t",
    index=False,
    encoding="utf-8-sig"
)

print("Multilingual dataset saved successfully!")
```

Prepare the labels and inspect the official split

```python
score_columns = ["agree", "disagree", "neutral"]

# Keep rows with all three annotation scores
labelled_df = multilingual_df.dropna(
    subset=score_columns
).copy()

# Convert scores into stance labels
labelled_df["stance"] = labelled_df[
    score_columns
].idxmax(axis=1)

labelled_df["stance"] = labelled_df["stance"].map({
    "agree": "Favour",
    "disagree": "Against",
    "neutral": "Neutral"
})

print("Labelled dataset shape:", labelled_df.shape)

print("\nOverall class distribution:")
print(labelled_df["stance"].value_counts())

print("\nClass distribution in official split:")
print(pd.crosstab(
    labelled_df["in_held_out_test"],
    labelled_df["stance"]
))
```
