Import libraries

```python
import pandas as pd
import numpy as np
import torch

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    TrainingArguments,
    Trainer
)

print("PyTorch version:", torch.__version__)
print("GPU available:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
```

Load our multilingual dataset

```python
multilingual_df = pd.read_csv("GWSD_Multilingual.tsv", sep="\t")

print("Dataset shape:", multilingual_df.shape)
print("\nColumns:")
print(multilingual_df.columns.tolist())

multilingual_df.head()
```

Prepare labels and official train/test split
We'll do exactly what we did in our baseline experiments:
- Remove 250 rows with missing aggregate stance scores.
- Convert the highest score into one of three labels.
- Preserve the official 200-row test set.
- Split the remaining 1,850 rows into 90% training and 10% validation.

```python
# Remove rows with missing stance scores
score_columns = ["agree", "disagree", "neutral"]

labelled_df = multilingual_df.dropna(
    subset=score_columns
).copy()

# Create stance labels
labelled_df["stance"] = labelled_df[score_columns].idxmax(axis=1)

labelled_df["stance"] = labelled_df["stance"].map({
    "agree": "Favour",
    "disagree": "Against",
    "neutral": "Neutral"
})

# Preserve official test split
train_val_df = labelled_df[
    labelled_df["in_held_out_test"] == False
].copy()

test_df = labelled_df[
    labelled_df["in_held_out_test"] == True
].copy()

# Split training data into training and validation
train_df, val_df = train_test_split(
    train_val_df,
    test_size=0.10,
    random_state=42,
    stratify=train_val_df["stance"]
)

# Reset indexes
train_df = train_df.reset_index(drop=True)
val_df = val_df.reset_index(drop=True)
test_df = test_df.reset_index(drop=True)

print("Training samples:", len(train_df))
print("Validation samples:", len(val_df))
print("Testing samples:", len(test_df))

print("\nTraining distribution:")
print(train_df["stance"].value_counts())

print("\nValidation distribution:")
print(val_df["stance"].value_counts())

print("\nTest distribution:")
print(test_df["stance"].value_counts())
```

Convert stance labels into numerical IDs
XLM-RoBERTa needs numerical labels instead of text labels.

We'll use this mapping:
Stance	Label ID
Against	0
Favour	1
Neutral	2

```python
label2id = {
    "Against": 0,
    "Favour": 1,
    "Neutral": 2
}

id2label = {
    0: "Against",
    1: "Favour",
    2: "Neutral"
}

train_df["label"] = train_df["stance"].map(label2id)
val_df["label"] = val_df["stance"].map(label2id)
test_df["label"] = test_df["stance"].map(label2id)

print("Training labels:")
print(train_df[["stance", "label"]].head())

print("\nUnique label IDs:", sorted(train_df["label"].unique()))
```

Load the XLM-RoBERTa tokenizer

```python
model_name = "FacebookAI/xlm-roberta-base"

tokenizer = AutoTokenizer.from_pretrained(model_name)

print("Tokenizer loaded successfully!")

sample_text = train_df["sentence"].iloc[0]

encoded_sample = tokenizer(
    sample_text,
    truncation=True,
    padding="max_length",
    max_length=128
)

print("\nOriginal sentence:")
print(sample_text)

print("\nToken IDs:")
print(encoded_sample["input_ids"])

print("\nDecoded sentence:")
print(tokenizer.decode(encoded_sample["input_ids"]))
```

A few things from output:
- <s> = beginning of sentence.
- </s> = end of sentence.
- <pad> = padding added to make sequences equal length.
- The Hugging Face warning is normal. Authentication is optional for downloading this public model.

Tokenize English dataset

```python
MAX_LENGTH = 128

train_encodings = tokenizer(
    train_df["sentence"].tolist(),
    truncation=True,
    padding="max_length",
    max_length=MAX_LENGTH
)

val_encodings = tokenizer(
    val_df["sentence"].tolist(),
    truncation=True,
    padding="max_length",
    max_length=MAX_LENGTH
)

test_encodings = tokenizer(
    test_df["sentence"].tolist(),
    truncation=True,
    padding="max_length",
    max_length=MAX_LENGTH
)

print("Training samples:", len(train_encodings["input_ids"]))
print("Validation samples:", len(val_encodings["input_ids"]))
print("Testing samples:", len(test_encodings["input_ids"]))

print("\nInput shape:", np.array(train_encodings["input_ids"]).shape)
print("Attention mask shape:", np.array(train_encodings["attention_mask"]).shape)
```

Convert the tokenized data into PyTorch Datasets.

```python
class StanceDataset(torch.utils.data.Dataset):
    def __init__(self, encodings, labels):
        self.encodings = encodings
        self.labels = labels.tolist()

    def __getitem__(self, idx):
        item = {
            key: torch.tensor(val[idx])
            for key, val in self.encodings.items()
        }
        item["labels"] = torch.tensor(
            self.labels[idx],
            dtype=torch.long
        )
        return item

    def __len__(self):
        return len(self.labels)


train_dataset = StanceDataset(train_encodings, train_df["label"])
val_dataset = StanceDataset(val_encodings, val_df["label"])
test_dataset = StanceDataset(test_encodings, test_df["label"])

print("Training dataset:", len(train_dataset))
print("Validation dataset:", len(val_dataset))
print("Testing dataset:", len(test_dataset))

print("\nSample keys:", train_dataset[0].keys())
print("Sample label:", train_dataset[0]["labels"])
```

# Load the XLM-RoBERTa model

```python
model = AutoModelForSequenceClassification.from_pretrained(
    model_name,
    num_labels=3,
    id2label=id2label,
    label2id=label2id
)

print("XLM-RoBERTa model loaded successfully!")
print("Number of labels:", model.config.num_labels)
print("GPU available:", torch.cuda.is_available())
```

Define evaluation metrics

```python
def compute_metrics(eval_pred):
    logits, labels = eval_pred

    predictions = np.argmax(logits, axis=-1)

    accuracy = accuracy_score(labels, predictions)

    precision = precision_score(
        labels, predictions,
        average="macro",
        zero_division=0
    )

    recall = recall_score(
        labels, predictions,
        average="macro",
        zero_division=0
    )

    f1 = f1_score(
        labels, predictions,
        average="macro",
        zero_division=0
    )

    return {
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1": f1
    }

print("Evaluation metrics function created successfully!")
```

Set training parameters

```python
training_args = TrainingArguments(
    output_dir="./xlmr_english_results",

    eval_strategy="epoch",
    save_strategy="epoch",

    learning_rate=2e-5,
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,

    num_train_epochs=3,
    weight_decay=0.01,

    load_best_model_at_end=True,
    metric_for_best_model="f1",
    greater_is_better=True,

    fp16=True,
    logging_steps=50,
    save_total_limit=1,

    report_to="none"
)

print("Training arguments configured successfully!")
```

Create the Trainer

```python
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=val_dataset,
    compute_metrics=compute_metrics,
    processing_class=tokenizer
)

print("Trainer created successfully!")
```

Train XLM-RoBERTa on English data

```python
training_output = trainer.train()
```

Evaluate on the official test dataset

```python
test_results = trainer.predict(test_dataset)

test_predictions = np.argmax(
    test_results.predictions,
    axis=1
)

test_true_labels = test_results.label_ids

print("English XLM-RoBERTa Test Results")
print("--------------------------------")

print("Accuracy:",
      accuracy_score(test_true_labels, test_predictions))

print("Precision:",
      precision_score(test_true_labels, test_predictions,
                      average="macro", zero_division=0))

print("Recall:",
      recall_score(test_true_labels, test_predictions,
                   average="macro", zero_division=0))

print("Macro F1:",
      f1_score(test_true_labels, test_predictions,
               average="macro", zero_division=0))

print("\nClassification Report:\n")

print(classification_report(
    test_true_labels,
    test_predictions,
    labels=[0, 1, 2],
    target_names=["Against", "Favour", "Neutral"],
    zero_division=0
))
```

Check Bengali text quality

```python
import re

def check_bengali_quality(text):
    text = str(text)

    bengali_chars = len(re.findall(r'[\u0980-\u09FF]', text))
    devanagari_chars = len(re.findall(r'[\u0900-\u097F]', text))
    latin_chars = len(re.findall(r'[a-zA-Z]', text))

    total = max(len(re.findall(r'\S', text)), 1)

    return pd.Series({
        "bengali_ratio": bengali_chars / total,
        "devanagari_ratio": devanagari_chars / total,
        "latin_ratio": latin_chars / total
    })

quality_df = train_df["sentence_bn"].apply(check_bengali_quality)

print("Average Bengali character ratio:",
      quality_df["bengali_ratio"].mean().round(3))

print("Average Devanagari character ratio:",
      quality_df["devanagari_ratio"].mean().round(3))

print("Average Latin character ratio:",
      quality_df["latin_ratio"].mean().round(3))

print("\nRows containing Devanagari characters:",
      (quality_df["devanagari_ratio"] > 0).sum())
```

## Generate clean Bengali translations using NLLB-200

```python
def translate_to_bengali(sentences, batch_size=32):
    translations = []
    for i in range(0, len(sentences), batch_size):
        batch = sentences[i : i + batch_size]
        inputs = bn_tokenizer(
            batch,
            return_tensors="pt",
            padding=True,
            truncation=True,
            max_length=128
        ).to(bn_model.device)

        with torch.no_grad():
            translated_tokens = bn_model.generate(
                **inputs,
                forced_bos_token_id=bengali_token_id,
                max_length=128,
                num_beams=4
            )

        decoded = bn_tokenizer.batch_decode(
            translated_tokens,
            skip_special_tokens=True
        )
        translations.extend(decoded)
    return translations

print("Translating train set (1,665 samples)...")
train_df["sentence_bn_clean"] = translate_to_bengali(train_df["sentence"].tolist())

print("Translating validation set (185 samples)...")
val_df["sentence_bn_clean"] = translate_to_bengali(val_df["sentence"].tolist())

print("Translating test set (200 samples)...")
test_df["sentence_bn_clean"] = translate_to_bengali(test_df["sentence"].tolist())

print("\nAll translations completed successfully!")
```

```python

import re
import pandas as pd

def check_bengali_quality(text):
    text = str(text)

    bengali_chars = len(re.findall(r'[\u0980-\u09FF]', text))
    devanagari_chars = len(re.findall(r'[\u0900-\u097F]', text))
    latin_chars = len(re.findall(r'[a-zA-Z]', text))

    total = max(len(re.findall(r'\S', text)), 1)

    return pd.Series({
        "bengali_ratio": bengali_chars / total,
        "devanagari_ratio": devanagari_chars / total,
        "latin_ratio": latin_chars / total
    })

quality_df = pd.concat([
    train_df["sentence_bn_clean"].apply(check_bengali_quality),
    val_df["sentence_bn_clean"].apply(check_bengali_quality),
    test_df["sentence_bn_clean"].apply(check_bengali_quality)
], ignore_index=True)

print("Average Bengali character ratio:",
      round(quality_df["bengali_ratio"].mean(), 4))

print("Average Devanagari character ratio:",
      round(quality_df["devanagari_ratio"].mean(), 4))

print("Average Latin character ratio:",
      round(quality_df["latin_ratio"].mean(), 4))

print("Rows containing Devanagari:",
      (quality_df["devanagari_ratio"] > 0).sum(), "out of", len(quality_df))
```

Recalculate Devanagari contamination

```python

# Combine all three splits in the same order
all_bn = pd.concat([
    train_df["sentence_bn_clean"],
    val_df["sentence_bn_clean"],
    test_df["sentence_bn_clean"]
], ignore_index=True)

# Recalculate directly from the actual translated text
dev_counts = all_bn.apply(
    lambda text: len(re.findall(r'[\u0900-\u097F]', str(text)))
)

print("Total sentences:", len(all_bn))
print("Sentences containing Devanagari:", (dev_counts > 0).sum())
print("Total Devanagari characters:", dev_counts.sum())

print("\nActual contaminated examples:")
for idx in dev_counts[dev_counts > 0].index[:10]:
    print("\nBengali:", all_bn.iloc[idx])
    print("Devanagari characters:", dev_counts.iloc[idx])
```

Identify the hidden characters

```python

# Inspect the exact Devanagari character and its Unicode code point

for idx in dev_counts[dev_counts > 0].index[:10]:
    text = str(all_bn.iloc[idx])

    hidden_chars = [
        (char, f"U+{ord(char):04X}",
         __import__("unicodedata").name(char, "UNKNOWN"))
        for char in text
        if '\u0900' <= char <= '\u097F'
    ]

    print("\nSentence:", text)
    print("Exact characters:", hidden_chars)
```

The character । (U+0964) is Devanagari Danda, which is being used as a sentence-ending punctuation mark.

It is not a Hindi word or a translation error. The Bengali sentences themselves look clean.

Bengali commonly uses । too, so we can safely retain it. Our earlier script-based quality check was misleading because it counted this punctuation as Devanagari contamination.

Correct the quality check

```python

# Count Devanagari letters, excluding the shared sentence-ending danda
def check_bengali_quality_corrected(text):
    text = str(text)

    # Exclude U+0964 from the Devanagari range
    devanagari_chars = len(
        re.findall(r'[\u0900-\u0963\u0965-\u097F]', text)
    )

    bengali_chars = len(re.findall(r'[\u0980-\u09FF]', text))
    latin_chars = len(re.findall(r'[a-zA-Z]', text))

    total = max(len(re.findall(r'\S', text)), 1)

    return pd.Series({
        "bengali_ratio": bengali_chars / total,
        "devanagari_ratio": devanagari_chars / total,
        "latin_ratio": latin_chars / total
    })

corrected_quality = all_bn.apply(check_bengali_quality_corrected)

print("Average Bengali ratio:",
      round(corrected_quality["bengali_ratio"].mean(), 4))

print("Average Devanagari ratio:",
      round(corrected_quality["devanagari_ratio"].mean(), 4))

print("Sentences with actual Devanagari letters:",
      (corrected_quality["devanagari_ratio"] > 0).sum())
```
